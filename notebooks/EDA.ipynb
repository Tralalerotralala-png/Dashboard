# Análisis exploratorio de anuncios de vehículos
El objetivo es explorar el kilometraje y el precio de los vehículos usados.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px

path = Path('../vehicles_us.csv')
if not path.exists():
    path = Path('vehicles_us.csv')
car_data = pd.read_csv(path)
car_data.head()
```

```python
car_data.info()
```

```python
car_data.describe()
```

```python
car_data.isna().sum()
```

## Distribución del kilometraje
Se excluyen los valores ausentes del odómetro para construir el histograma.

```python
fig = px.histogram(car_data.dropna(subset=['odometer']), x='odometer', nbins=30, title='Distribución del kilometraje', labels={'odometer':'Odómetro (millas)'})
fig.show()
```

## Relación entre precio y kilometraje
El gráfico de dispersión permite comparar los precios y el kilometraje por tipo de vehículo.

```python
fig = px.scatter(car_data.dropna(subset=['odometer','price','type']), x='odometer', y='price', color='type', title='Precio frente al kilometraje', labels={'odometer':'Odómetro (millas)','price':'Precio (USD)','type':'Tipo de vehículo'}, opacity=0.55)
fig.show()
```

```python
car_data[['odometer', 'price']].corr()
```

## Conclusión
El histograma permite identificar los rangos de kilometraje más frecuentes, mientras que el gráfico de dispersión ayuda a explorar la relación entre el uso del vehículo y su precio anunciado. Los valores ausentes se excluyen únicamente de las visualizaciones que los requieren.
